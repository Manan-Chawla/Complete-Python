```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
data={
    "text":[
        "Congrats you won a iphone",
        "call number to get prize",
        "are we meeting tomorrow for work?",
        "you have been selected for a lottery",
        "let's catch up in evening"
    ],
    "labels":[1,1,0,1,0]
}
```

```python
df=pd.DataFrame(data)
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    df["text"],
    df["labels"],
    test_size=0.3,
    random_state=42
)
```

```python
vectorizer=CountVectorizer()
x_train_vec=vectorizer.fit_transform(x_train)
x_test_vec=vectorizer.transform(x_test)
```

```python
model=MultinomialNB()
model.fit(x_train_vec,y_train)

pred=model.predict(x_test_vec)
print("accuracy score is : ",accuracy_score(y_test,pred))
```

```python
msg=["You win a cash prize of 8 millio dollars in india :)"]
print("Prediction : ",model.predict(vectorizer.transform(msg)))
```
