```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
data = {
    "age": [18, 22, 25, 30, 35, 40, 45, 50],
    "buy": [0, 0, 0, 1, 1, 1, 1, 0]
}
```

```python
df = pd.DataFrame(data)

X = df[["age"]]
y = df["buy"]
```

```python
model_entropy_1 = DecisionTreeClassifier(criterion="entropy")
model_entropy_1.fit(X, y)

print(model_entropy_1.predict([[24]]))
print(model_entropy_1.predict([[38]]))
```
