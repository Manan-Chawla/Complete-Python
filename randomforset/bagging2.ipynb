```python
from sklearn.ensemble import BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
```

```python
X = [[500], [600], [700], [800], [900]]
y = [25, 30, 35, 45, 55]

bag_reg = BaggingRegressor(
    estimator=DecisionTreeRegressor(),
    n_estimators=15
)

bag_reg.fit(X, y)
print(bag_reg.predict([[750]]))
```
