```python
from sklearn.ensemble import VotingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
```

```python
X = [[10], [20], [30], [40]]
y = [100, 200, 300, 400]

vr = VotingRegressor([
    ('lr', LinearRegression()),
    ('dt', DecisionTreeRegressor())
])

vr.fit(X, y)
print(vr.predict([[25]]))
```
