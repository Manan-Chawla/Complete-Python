```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
X = [[1], [2], [3], [4], [5], [6]]
y = [0, 0, 0, 1, 1, 1]

bag1 = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=20
)

bag1.fit(X, y)
print(bag1.predict([[4]]))
```
