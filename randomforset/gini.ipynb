```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
data={
    "study_hours":[1,2,3,4,5,6],
    "pass_fail":[0,0,0,1,1,0]
}
```

```python
df = pd.DataFrame(data)

X = df[["study_hours"]]
y = df["pass_fail"]
```

```python
model_gini_1 = DecisionTreeClassifier(criterion="gini")
model_gini_1.fit(X, y)


print(model_gini_1.predict([[3]]))  
print(model_gini_1.predict([[6]]))
```
