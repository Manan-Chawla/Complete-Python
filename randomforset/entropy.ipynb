```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
data = {
    "email_length": [50, 70, 100, 150, 200, 250, 300, 350],
    "spam": [1, 1, 1, 0, 0, 0, 0, 0]
}
```

```python
df = pd.DataFrame(data)

X = df[["email_length"]]
y = df["spam"]
```

```python
model_entropy_2 = DecisionTreeClassifier(criterion="entropy")
model_entropy_2.fit(X, y)

print(model_entropy_2.predict([[80]]))
print(model_entropy_2.predict([[280]]))
```
