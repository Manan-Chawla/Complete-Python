```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
data={
    "income": [2, 3, 4, 5, 6, 7, 8, 9],
    "loan_approved": [0, 0, 0, 1, 1, 1, 1, 1]
}
```

```python
df = pd.DataFrame(data)

X = df[["income"]]
y = df["loan_approved"]
```

```python
model_gini_2 = DecisionTreeClassifier(criterion="gini")
model_gini_2.fit(X, y)

print(model_gini_2.predict([[4]]))
print(model_gini_2.predict([[7]])) 
```
